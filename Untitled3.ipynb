```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

print("TensorFlow sürümü:", tf.__version__)
```

```python
# Fashion-MNIST: 60.000 eğitim ve 10.000 test görüntüsü
(x_train_raw, y_train), (x_test_raw, y_test) = keras.datasets.fashion_mnist.load_data()

class_names = [
    "Tişört/Üst", "Pantolon", "Kazak", "Elbise", "Mont",
    "Sandalet", "Gömlek", "Spor ayakkabı", "Çanta", "Bilekte bot"
]

print("Eğitim görüntüleri:", x_train_raw.shape)
print("Eğitim etiketleri:", y_train.shape)
print("Test görüntüleri:", x_test_raw.shape)
print("Test etiketleri:", y_test.shape)
print("Piksel aralığı:", x_train_raw.min(), "-", x_train_raw.max())
```

```python
# Örnek görüntüleri ve etiketlerini göster
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train_raw[i], cmap="gray")
    plt.title(class_names[y_train[i]])
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Normalize et
x_train = x_train_raw.astype("float32") / 255.0
x_test = x_test_raw.astype("float32") / 255.0

# 28x28 görüntüyü 784 boyutlu vektör haline getir
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)

print("Yeni eğitim şekli:", x_train.shape)
print("Yeni test şekli:", x_test.shape)
print("Normalize edilmiş piksel aralığı:", x_train.min(), "-", x_train.max())
```

```python
def build_model():
    model = keras.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.2),
        layers.Dense(128, activation="relu"),
        layers.Dropout(0.2),
        layers.Dense(10, activation="softmax")
    ])
    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

model = build_model()
model.summary()
```

```python
EPOCHS_20 = 20
history_20 = model.fit(
    x_train, y_train,
    epochs=EPOCHS_20,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)
```

```python
def plot_history(history, title="Eğitim geçmişi"):
    epochs = range(1, len(history.history["loss"]) + 1)

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(epochs, history.history["loss"], label="Eğitim loss")
    plt.plot(epochs, history.history["val_loss"], label="Doğrulama loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"{title} - Loss")
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.subplot(1, 2, 2)
    plt.plot(epochs, history.history["accuracy"], label="Eğitim accuracy")
    plt.plot(epochs, history.history["val_accuracy"], label="Doğrulama accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(f"{title} - Accuracy")
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

plot_history(history_20, "20 Epoch")
```

```python
test_loss_20, test_acc_20 = model.evaluate(x_test, y_test, verbose=0)
print(f"20 epoch test loss: {test_loss_20:.4f}")
print(f"20 epoch test accuracy: {test_acc_20:.4f} ({test_acc_20 * 100:.2f}%)")
```

```python
# Test kümesinden örnekler için tahmin üret
sample_count = 15
probabilities = model.predict(x_test[:sample_count], verbose=0)
predicted_labels = np.argmax(probabilities, axis=1)

plt.figure(figsize=(14, 7))
for i in range(sample_count):
    plt.subplot(3, 5, i + 1)
    plt.imshow(x_test_raw[i], cmap="gray")
    true_name = class_names[y_test[i]]
    pred_name = class_names[predicted_labels[i]]
    color = "green" if y_test[i] == predicted_labels[i] else "red"
    plt.title(f"Gerçek: {true_name}\nTahmin: {pred_name}", color=color, fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Sınıflandırma raporu
y_pred_all = np.argmax(model.predict(x_test, verbose=0), axis=1)
print(classification_report(y_test, y_pred_all, target_names=class_names, digits=3))
```

```python
cm = confusion_matrix(y_test, y_pred_all)
fig, ax = plt.subplots(figsize=(10, 8))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
disp.plot(ax=ax, cmap="Blues", xticks_rotation=45, colorbar=False)
plt.title("Fashion-MNIST - Karışıklık Matrisi")
plt.tight_layout()
plt.show()
```

```python
EPOCHS_30 = 30
model_30 = build_model()

history_30 = model_30.fit(
    x_train, y_train,
    epochs=EPOCHS_30,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

plot_history(history_30, "30 Epoch")
```

```python
test_loss_30, test_acc_30 = model_30.evaluate(x_test, y_test, verbose=0)
print(f"30 epoch test loss: {test_loss_30:.4f}")
print(f"30 epoch test accuracy: {test_acc_30:.4f} ({test_acc_30 * 100:.2f}%)")
```

```python
comparison = {
    "Epoch": [20, 30],
    "Test Loss": [test_loss_20, test_loss_30],
    "Test Accuracy": [test_acc_20, test_acc_30]
}
for i in range(2):
    print(f'{comparison["Epoch"][i]} epoch -> '
          f'Loss: {comparison["Test Loss"][i]:.4f}, '
          f'Accuracy: {comparison["Test Accuracy"][i] * 100:.2f}%')

accuracy_difference = (test_acc_30 - test_acc_20) * 100
print(f"Accuracy farkı (30 - 20 epoch): {accuracy_difference:+.2f} yüzde puan")
```

```python
# Doğrulama accuracy eğrilerini tek grafikte karşılaştır
plt.figure(figsize=(9, 5))
plt.plot(range(1, 21), history_20.history["val_accuracy"], label="20 epoch modeli")
plt.plot(range(1, 31), history_30.history["val_accuracy"], label="30 epoch modeli")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.title("20 ve 30 Epoch Doğrulama Accuracy Karşılaştırması")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
